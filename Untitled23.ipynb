```python
# Assignment 4: CNN Layer Comparison
# Name: [Your Name Here]
# Student ID: [Your Student ID Here]

import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
import numpy as np
import time
from datetime import datetime

# Check if CUDA is available for faster training
device = torch.device('cuda:0' if torch.cuda.is_available() else 'cpu')
print(f'Using device: {device}')

# ===== DATA LOADING AND PREPROCESSING =====
print("Loading CIFAR-10 dataset...")

# Standard transforms (keeping it simple as in typical assignment)
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))  # Standard normalization
])

# Standard batch size
batch_size = 4  # Using smaller batch size as typically used in assignments

trainset = torchvision.datasets.CIFAR10(root='./data', train=True,
                                        download=True, transform=transform)
trainloader = torch.utils.data.DataLoader(trainset, batch_size=batch_size,
                                          shuffle=True, num_workers=2)

testset = torchvision.datasets.CIFAR10(root='./data', train=False,
                                       download=True, transform=transform)
testloader = torch.utils.data.DataLoader(testset, batch_size=batch_size,
                                         shuffle=False, num_workers=2)

classes = ('plane', 'car', 'bird', 'cat', 'deer', 'dog', 'frog', 'horse', 'ship', 'truck')

# ===== VISUALIZATION FUNCTION =====
def imshow(img, title=None):
    """Display image with optional title"""
    img = img / 2 + 0.5  # unnormalize
    npimg = img.numpy()
    plt.figure(figsize=(8, 4))
    plt.imshow(np.transpose(npimg, (1, 2, 0)))
    if title:
        plt.title(title)
    plt.axis('off')
    plt.show()

# Show sample images
print("\nSample training images:")
dataiter = iter(trainloader)
images, labels = next(dataiter)
imshow(torchvision.utils.make_grid(images),
       title=' '.join(f'{classes[labels[j]]:5s}' for j in range(4)))

# ===== ORIGINAL CNN ARCHITECTURE =====
class OriginalCNN(nn.Module):
    """Original CNN with all layers: Conv2d -> ReLU -> MaxPool2d -> Conv2d -> ReLU -> MaxPool2d -> FC"""
    def __init__(self):
        super().__init__()
        # First convolutional block
        self.conv1 = nn.Conv2d(3, 6, 5)       # Standard LeNet-style architecture
        self.pool = nn.MaxPool2d(2, 2)
        # Second convolutional block
        self.conv2 = nn.Conv2d(6, 16, 5)      # Standard LeNet-style architecture
        # Fully connected layers
        self.fc1 = nn.Linear(16 * 5 * 5, 120)
        self.fc2 = nn.Linear(120, 84)
        self.fc3 = nn.Linear(84, 10)

    def forward(self, x):
        x = self.pool(F.relu(self.conv1(x)))  # Conv1 -> ReLU -> Pool
        x = self.pool(F.relu(self.conv2(x)))  # Conv2 -> ReLU -> Pool
        x = torch.flatten(x, 1)
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))
        x = self.fc3(x)
        return x

# ===== MODIFIED CNN ARCHITECTURE (REMOVED ONE LAYER SET) =====
class ModifiedCNN(nn.Module):
    """Modified CNN with one conv layer removed: Conv2d -> ReLU -> MaxPool2d -> FC"""
    def __init__(self):
        super().__init__()
        # Only first convolutional block (second block removed)
        self.conv1 = nn.Conv2d(3, 6, 5)       # Same as original first layer
        self.pool = nn.MaxPool2d(2, 2)
        # Fully connected layers (adjusted input size)
        self.fc1 = nn.Linear(6 * 14 * 14, 120)  # Adjusted for larger feature maps
        self.fc2 = nn.Linear(120, 84)
        self.fc3 = nn.Linear(84, 10)

    def forward(self, x):
        x = self.pool(F.relu(self.conv1(x)))  # Only one Conv -> ReLU -> Pool
        x = torch.flatten(x, 1)
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))
        x = self.fc3(x)
        return x

# ===== TRAINING FUNCTION =====
def train_model(model, model_name, epochs=2):  # Standard 2 epochs for assignment
    """Train the model and return training history"""
    print(f"\n===== Training {model_name} =====")

    model = model.to(device)
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.SGD(model.parameters(), lr=0.001, momentum=0.9)  # Standard SGD

    train_losses = []
    start_time = time.time()

    for epoch in range(epochs):
        running_loss = 0.0
        epoch_start = time.time()

        model.train()
        for i, data in enumerate(trainloader, 0):
            inputs, labels = data[0].to(device), data[1].to(device)

            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()

            running_loss += loss.item()

            # Print progress every 2000 mini-batches
            if i % 2000 == 1999:
                print(f'[{epoch + 1}, {i + 1:5d}] loss: {running_loss / 2000:.3f}')
                running_loss = 0.0

        epoch_time = time.time() - epoch_start
        print(f'Epoch {epoch + 1} completed in {epoch_time:.1f}s')

    total_time = time.time() - start_time
    print(f'{model_name} training completed in {total_time:.1f} seconds')

    return model, total_time

# ===== TESTING FUNCTION =====
def test_model(model, model_name):
    """Test the model and return detailed results"""
    print(f"\n===== Testing {model_name} =====")

    model.eval()
    correct = 0
    total = 0
    class_correct = list(0. for i in range(10))
    class_total = list(0. for i in range(10))

    with torch.no_grad():
        for data in testloader:
            images, labels = data[0].to(device), data[1].to(device)
            outputs = model(images)
            _, predicted = torch.max(outputs, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()

            # Per-class accuracy
            c = (predicted == labels).squeeze()
            for i in range(labels.size(0)):
                label = labels[i]
                class_correct[label] += c[i].item()
                class_total[label] += 1

    overall_accuracy = 100 * correct / total
    print(f'Overall Test Accuracy: {overall_accuracy:.2f}%')

    print('\nPer-class accuracies:')
    class_accuracies = {}
    for i in range(10):
        if class_total[i] > 0:
            acc = 100 * class_correct[i] / class_total[i]
            class_accuracies[classes[i]] = acc
            print(f'{classes[i]:10s}: {acc:.1f}%')

    return overall_accuracy, class_accuracies

# ===== VISUALIZATION FUNCTION =====
def visualize_predictions(original_model, modified_model):
    """Show sample predictions from both models"""
    print("\n===== Sample Predictions Comparison =====")

    dataiter = iter(testloader)
    images, labels = next(dataiter)
    images, labels = images.to(device), labels.to(device)

    # Get predictions
    with torch.no_grad():
        orig_outputs = original_model(images)
        mod_outputs = modified_model(images)

        _, orig_predicted = torch.max(orig_outputs, 1)
        _, mod_predicted = torch.max(mod_outputs, 1)

    # Display results
    print(f"{'Image':<8} {'True':<10} {'Original':<10} {'Modified':<10} {'Orig✓':<6} {'Mod✓':<6}")
    print("-" * 60)

    for i in range(min(4, len(labels))):  # Show first 4 images
        true_label = classes[labels[i]]
        orig_pred = classes[orig_predicted[i]]
        mod_pred = classes[mod_predicted[i]]
        orig_correct = "✓" if orig_predicted[i] == labels[i] else "✗"
        mod_correct = "✓" if mod_predicted[i] == labels[i] else "✗"

        print(f"{i+1:<8} {true_label:<10} {orig_pred:<10} {mod_pred:<10} {orig_correct:<6} {mod_correct:<6}")

# ===== MAIN EXECUTION =====
print(f"\n{'='*60}")
print("ASSIGNMENT 4: CNN LAYER COMPARISON")
print(f"{'='*60}")
print(f"Started at: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")

# Initialize models
original_model = OriginalCNN()
modified_model = ModifiedCNN()

# Display model architectures
print(f"\nOriginal CNN Architecture:")
print(f"Parameters: {sum(p.numel() for p in original_model.parameters()):,}")
print(original_model)

print(f"\nModified CNN Architecture (one conv layer removed):")
print(f"Parameters: {sum(p.numel() for p in modified_model.parameters()):,}")
print(modified_model)

# Train both models
original_trained, orig_time = train_model(original_model, "Original CNN")
modified_trained, mod_time = train_model(modified_model, "Modified CNN")

# Test both models
orig_test_acc, orig_class_acc = test_model(original_trained, "Original CNN")
mod_test_acc, mod_class_acc = test_model(modified_trained, "Modified CNN")

# Show sample predictions
visualize_predictions(original_trained, modified_trained)

# ===== FINAL RESULTS SUMMARY =====
print(f"\n{'='*60}")
print("FINAL RESULTS SUMMARY")
print(f"{'='*60}")

print(f"Original CNN (Full Architecture):")
print(f"  - Test Accuracy: {orig_test_acc:.2f}%")
print(f"  - Training Time: {orig_time:.1f} seconds")
print(f"  - Parameters: {sum(p.numel() for p in OriginalCNN().parameters()):,}")

print(f"\nModified CNN (One Conv Layer Removed):")
print(f"  - Test Accuracy: {mod_test_acc:.2f}%")
print(f"  - Training Time: {mod_time:.1f} seconds")
print(f"  - Parameters: {sum(p.numel() for p in ModifiedCNN().parameters()):,}")

print(f"\nComparison:")
accuracy_diff = orig_test_acc - mod_test_acc
print(f"  - Accuracy Difference: {accuracy_diff:.2f}% {'(Original better)' if accuracy_diff > 0 else '(Modified better)'}")
print(f"  - Speed Difference: {abs(orig_time - mod_time):.1f}s {'(Modified faster)' if mod_time < orig_time else '(Original faster)'}")

orig_params = sum(p.numel() for p in OriginalCNN().parameters())
mod_params = sum(p.numel() for p in ModifiedCNN().parameters())
param_reduction = ((orig_params - mod_params) / orig_params * 100)
print(f"  - Parameter Reduction: {param_reduction:.1f}%")

# ===== MY INTERPRETATION OF RESULTS =====
print(f"\n{'='*60}")
print("MY INTERPRETATION OF RESULTS")
print(f"{'='*60}")

interpretation = f"""
Based on the experimental results above, here are my observations and interpretations:

1. ACCURACY IMPACT:
   - The removal of one convolutional layer (Conv2d + ReLU + MaxPool2d) resulted in
     {accuracy_diff:.2f}% change in test accuracy.
   - {"The original model performed better, indicating that the second convolutional layer was important for feature extraction and pattern recognition." if accuracy_diff > 0 else "Surprisingly, the modified model performed better, suggesting that the original architecture might have been prone to overfitting or the second conv layer was redundant."}

2. TRAINING EFFICIENCY:
   - The modified CNN trained {'faster' if mod_time < orig_time else 'slower'} than the original, completing in {abs(orig_time - mod_time):.1f}s less time.
   - This is {'expected' if mod_time < orig_time else 'unexpected'} because {'fewer' if param_reduction > 0 else 'more'} parameters means {'less' if param_reduction > 0 else 'more'} computation per forward pass.

3. MODEL COMPLEXITY:
   - Parameter reduction of {param_reduction:.1f}% shows {'significant model simplification' if param_reduction > 0 else 'increased model complexity'}.
   - {'Fewer parameters generally mean faster training, less memory usage, and reduced risk of overfitting, but potentially lower representational capacity.' if param_reduction > 0 else 'More parameters allow for greater representational capacity but increase computational cost and overfitting risk.'}

4. FEATURE LEARNING:
   - The original CNN has two convolutional layers that can learn hierarchical features:
     * First layer: Low-level features (edges, corners)
     * Second layer: Mid-level features (shapes, textures)
   - The modified CNN relies on just one conv layer, limiting its ability to learn complex hierarchical features.
   - This explains why {'the original model achieved better accuracy - deeper feature extraction was beneficial' if accuracy_diff > 0 else 'the modified model performed similarly or better - the dataset might not require deep hierarchical features'}.

5. PRACTICAL IMPLICATIONS:
   - For applications requiring {'high accuracy' if accuracy_diff > 0 else 'computational efficiency'}, the {'original' if accuracy_diff > 0 else 'modified'} model would be preferred.
   - The trade-off between model complexity and performance is evident here.
   - In resource-constrained environments (mobile devices, embedded systems), the {'modified' if param_reduction > 0 else 'original'} model might be more suitable despite {'any accuracy loss' if accuracy_diff > 0 and param_reduction > 0 else 'increased complexity'}.

6. ARCHITECTURE INSIGHTS:
   - This experiment demonstrates that not all layers contribute equally to performance.
   - For simple datasets like CIFAR-10, {'excessive depth might lead to overfitting' if accuracy_diff < 0 else 'multiple convolutional layers help capture important patterns'}.
   - The optimal architecture depends on the complexity of the dataset and the available computational resources.

CONCLUSION:
The experiment demonstrates the importance of each layer in CNN architecture and shows
how architectural choices affect both performance and computational efficiency.
{'Removing the second convolutional layer resulted in reduced accuracy but faster training, highlighting the trade-off between model capacity and efficiency.' if accuracy_diff > 0 else 'Removing the second convolutional layer improved or maintained performance while reducing training time, suggesting the original architecture may have been unnecessarily complex for this task.'}

This analysis reinforces the importance of careful architecture design and the need to
balance model complexity with performance requirements for specific applications.
"""

print(interpretation)
```

```python
# Assignment 4: CNN Layer Comparison
# Name: Kirtirajsinh Atodariya
# Student ID: N01620157

import torch
import torchvision
import matplotlib.pyplot as plt
import numpy as np
import time
from datetime import datetime

# The below two lines are optional and are just there to avoid any SSL
# related errors while downloading the CIFAR-10 dataset
import ssl
ssl._create_default_https_context = ssl._create_unverified_context

# Defining plotting settings
plt.rcParams['figure.figsize'] = 14, 6

# Check if CUDA is available for faster training
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'Using device: {device}')

# ===== DATA LOADING AND PREPROCESSING =====
print("Loading CIFAR-10 dataset...")

# Initializing normalizing transform for the dataset (from GeeksforGeeks)
normalize_transform = torchvision.transforms.Compose([
    torchvision.transforms.ToTensor(),
    torchvision.transforms.Normalize(mean=(0.5, 0.5, 0.5),
                                   std=(0.5, 0.5, 0.5))
])

# Downloading the CIFAR10 dataset into train and test sets
train_dataset = torchvision.datasets.CIFAR10(
    root="./CIFAR10/train", train=True,
    transform=normalize_transform,
    download=True)
test_dataset = torchvision.datasets.CIFAR10(
    root="./CIFAR10/test", train=False,
    transform=normalize_transform,
    download=True)

# Generating data loaders from the corresponding datasets
batch_size = 128
train_loader = torch.utils.data.DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
test_loader = torch.utils.data.DataLoader(test_dataset, batch_size=batch_size, shuffle=False)

# Classes in CIFAR-10
classes = ['airplane', 'automobile', 'bird', 'cat', 'deer', 'dog', 'frog', 'horse', 'ship', 'truck']

# ===== VISUALIZATION FUNCTIONS =====
def show_sample_images():
    """Plotting 25 images from the 1st batch"""
    dataiter = iter(train_loader)
    images, labels = next(dataiter)
    plt.imshow(np.transpose(torchvision.utils.make_grid(
        images[:25], normalize=True, padding=1, nrow=5).numpy(), (1, 2, 0)))
    plt.axis('off')
    plt.title('Sample training images')
    plt.show()

def plot_class_distribution():
    """Plotting class distribution of the dataset"""
    # Iterating over the training dataset and storing the target class for each sample
    class_labels = []
    for batch_idx, data in enumerate(train_loader, 0):
        x, y = data
        class_labels.extend(y.tolist())

    # Calculating the unique classes and the respective counts and plotting them
    unique, counts = np.unique(class_labels, return_counts=True)
    names = list(test_dataset.class_to_idx.keys())
    plt.figure(figsize=(10, 6))
    plt.bar(names, counts)
    plt.xlabel("Target Classes")
    plt.ylabel("Number of training instances")
    plt.title("Class Distribution")
    plt.xticks(rotation=45)
    plt.show()

# Show sample data
show_sample_images()
plot_class_distribution()

# ===== ORIGINAL CNN ARCHITECTURE (FROM GEEKSFORGEEKS) =====
class OriginalCNN(torch.nn.Module):
    """Original CNN from GeeksforGeeks practice program"""
    def __init__(self):
        super().__init__()
        self.model = torch.nn.Sequential(
            # Input = 3 x 32 x 32, Output = 32 x 32 x 32
            torch.nn.Conv2d(in_channels=3, out_channels=32, kernel_size=3, padding=1),
            torch.nn.ReLU(),
            # Input = 32 x 32 x 32, Output = 32 x 16 x 16
            torch.nn.MaxPool2d(kernel_size=2),
            # Input = 32 x 16 x 16, Output = 64 x 16 x 16
            torch.nn.Conv2d(in_channels=32, out_channels=64, kernel_size=3, padding=1),
            torch.nn.ReLU(),
            # Input = 64 x 16 x 16, Output = 64 x 8 x 8
            torch.nn.MaxPool2d(kernel_size=2),
            # Input = 64 x 8 x 8, Output = 64 x 8 x 8
            torch.nn.Conv2d(in_channels=64, out_channels=64, kernel_size=3, padding=1),
            torch.nn.ReLU(),
            # Input = 64 x 8 x 8, Output = 64 x 4 x 4
            torch.nn.MaxPool2d(kernel_size=2),
            torch.nn.Flatten(),
            torch.nn.Linear(64*4*4, 512),
            torch.nn.ReLU(),
            torch.nn.Linear(512, 10)
        )

    def forward(self, x):
        return self.model(x)

# ===== MODIFIED CNN ARCHITECTURE (REMOVED ONE CONV+RELU+MAXPOOL SET) =====
class ModifiedCNN(torch.nn.Module):
    """Modified CNN with second convolutional block removed"""
    def __init__(self):
        super().__init__()
        self.model = torch.nn.Sequential(
            # Input = 3 x 32 x 32, Output = 32 x 32 x 32
            torch.nn.Conv2d(in_channels=3, out_channels=32, kernel_size=3, padding=1),
            torch.nn.ReLU(),
            # Input = 32 x 32 x 32, Output = 32 x 16 x 16
            torch.nn.MaxPool2d(kernel_size=2),
            # REMOVED: Second Conv2d + ReLU + MaxPool2d block
            # Input = 32 x 16 x 16, Output = 64 x 16 x 16
            torch.nn.Conv2d(in_channels=32, out_channels=64, kernel_size=3, padding=1),
            torch.nn.ReLU(),
            # Input = 64 x 16 x 16, Output = 64 x 8 x 8
            torch.nn.MaxPool2d(kernel_size=2),
            torch.nn.Flatten(),
            torch.nn.Linear(64*8*8, 512),  # Adjusted input size
            torch.nn.ReLU(),
            torch.nn.Linear(512, 10)
        )

    def forward(self, x):
        return self.model(x)

# ===== TRAINING FUNCTION =====
def train_model(model, model_name):
    """Train the model following GeeksforGeeks approach"""
    print(f"\n===== Training {model_name} =====")

    # Selecting the appropriate training device
    model = model.to(device)

    # Defining the model hyper parameters (from GeeksforGeeks)
    num_epochs = 10  # Reduced from 50 for faster execution
    learning_rate = 0.001
    weight_decay = 0.01
    criterion = torch.nn.CrossEntropyLoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate, weight_decay=weight_decay)

    # Training process begins
    train_loss_list = []
    start_time = time.time()

    for epoch in range(num_epochs):
        print(f'Epoch {epoch+1}/{num_epochs}:', end=' ')
        train_loss = 0

        # Iterating over the training dataset in batches
        model.train()
        for i, (images, labels) in enumerate(train_loader):
            # Extracting images and target labels for the batch being iterated
            images = images.to(device)
            labels = labels.to(device)

            # Calculating the model output and the cross entropy loss
            outputs = model(images)
            loss = criterion(outputs, labels)

            # Updating weights according to calculated loss
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()
            train_loss += loss.item()

        # Printing loss for each epoch
        train_loss_list.append(train_loss/len(train_loader))
        print(f"Training loss = {train_loss_list[-1]:.4f}")

    total_time = time.time() - start_time
    print(f'{model_name} training completed in {total_time:.1f} seconds')

    # Plotting loss for all epochs
    plt.figure(figsize=(10, 6))
    plt.plot(range(1, num_epochs+1), train_loss_list)
    plt.xlabel("Number of epochs")
    plt.ylabel("Training loss")
    plt.title(f"Training Loss - {model_name}")
    plt.grid(True)
    plt.show()

    return model, total_time

# ===== TESTING FUNCTION =====
def test_model(model, model_name):
    """Test the model following GeeksforGeeks approach"""
    print(f"\n===== Testing {model_name} =====")

    test_acc = 0
    model.eval()
    class_correct = list(0. for i in range(10))
    class_total = list(0. for i in range(10))

    with torch.no_grad():
        # Iterating over the test dataset in batches
        for i, (images, labels) in enumerate(test_loader):
            images = images.to(device)
            y_true = labels.to(device)

            # Calculating outputs for the batch being iterated
            outputs = model(images)

            # Calculate prediction labels from models
            _, y_pred = torch.max(outputs.data, 1)

            # Comparing predicted and true labels
            test_acc += (y_pred == y_true).sum().item()

            # Per-class accuracy calculation
            c = (y_pred == y_true).squeeze()
            for j in range(y_true.size(0)):
                label = y_true[j]
                class_correct[label] += c[j].item()
                class_total[label] += 1

    overall_accuracy = 100 * test_acc / len(test_dataset)
    print(f"Test set accuracy = {overall_accuracy:.2f}%")

    # Per-class accuracies
    print('\nPer-class accuracies:')
    class_accuracies = {}
    for i in range(10):
        if class_total[i] > 0:
            acc = 100 * class_correct[i] / class_total[i]
            class_accuracies[classes[i]] = acc
            print(f'{classes[i]:10s}: {acc:.1f}%')

    return overall_accuracy, class_accuracies

# ===== PREDICTION VISUALIZATION =====
def visualize_predictions(model, model_name):
    """Generate predictions for sample images following GeeksforGeeks approach"""
    print(f"\n===== Sample Predictions for {model_name} =====")

    # Get some test images
    dataiter = iter(test_loader)
    images, y_true = next(dataiter)
    images = images.to(device)
    y_true = y_true.to(device)

    # Generate predictions
    model.eval()
    with torch.no_grad():
        outputs = model(images)
        _, y_pred = torch.max(outputs.data, 1)

    # Generating predictions for 'num_images' amount of images from the batch
    num_images = 5
    y_true_name = [classes[y_true[idx]] for idx in range(num_images)]
    y_pred_name = [classes[y_pred[idx]] for idx in range(num_images)]

    # Generating the title for the plot
    title = f"Actual: {y_true_name}\nPredicted: {y_pred_name}"

    # Finally plotting the images with their actual and predicted labels in the title
    plt.figure(figsize=(12, 8))
    plt.imshow(np.transpose(torchvision.utils.make_grid(
        images[:num_images].cpu(), normalize=True, padding=1).numpy(), (1, 2, 0)))
    plt.title(title)
    plt.axis("off")
    plt.show()

    # Show comparison table
    print(f"{'Image':<8} {'True':<12} {'Predicted':<12} {'Correct':<8}")
    print("-" * 50)
    for i in range(num_images):
        correct = "✓" if y_pred[i] == y_true[i] else "✗"
        print(f"{i+1:<8} {y_true_name[i]:<12} {y_pred_name[i]:<12} {correct:<8}")

# ===== MAIN EXECUTION =====
print(f"\n{'='*60}")
print("ASSIGNMENT 4: CNN LAYER COMPARISON")
print("Following GeeksforGeeks Practice Program")
print(f"{'='*60}")
print(f"Started at: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")

# Initialize models
original_model = OriginalCNN()
modified_model = ModifiedCNN()

# Display model architectures
print(f"\nOriginal CNN Architecture (GeeksforGeeks):")
print(f"Parameters: {sum(p.numel() for p in original_model.parameters()):,}")
print(original_model)

print(f"\nModified CNN Architecture (one conv block removed):")
print(f"Parameters: {sum(p.numel() for p in modified_model.parameters()):,}")
print(modified_model)

# Train both models
print(f"\n{'='*60}")
print("TRAINING PHASE")
print(f"{'='*60}")

original_trained, orig_time = train_model(original_model, "Original CNN")
modified_trained, mod_time = train_model(modified_model, "Modified CNN")

# Test both models
print(f"\n{'='*60}")
print("TESTING PHASE")
print(f"{'='*60}")

orig_test_acc, orig_class_acc = test_model(original_trained, "Original CNN")
mod_test_acc, mod_class_acc = test_model(modified_trained, "Modified CNN")

# Show sample predictions
print(f"\n{'='*60}")
print("PREDICTION VISUALIZATION")
print(f"{'='*60}")

visualize_predictions(original_trained, "Original CNN")
visualize_predictions(modified_trained, "Modified CNN")

# ===== FINAL RESULTS SUMMARY =====
print(f"\n{'='*60}")
print("FINAL RESULTS SUMMARY")
print(f"{'='*60}")

print(f"Original CNN (Full Architecture):")
print(f"  - Test Accuracy: {orig_test_acc:.2f}%")
print(f"  - Training Time: {orig_time:.1f} seconds")
print(f"  - Parameters: {sum(p.numel() for p in OriginalCNN().parameters()):,}")

print(f"\nModified CNN (One Conv Layer Removed):")
print(f"  - Test Accuracy: {mod_test_acc:.2f}%")
print(f"  - Training Time: {mod_time:.1f} seconds")
print(f"  - Parameters: {sum(p.numel() for p in ModifiedCNN().parameters()):,}")

print(f"\nComparison:")
accuracy_diff = orig_test_acc - mod_test_acc
print(f"  - Accuracy Difference: {accuracy_diff:.2f}% {'(Original better)' if accuracy_diff > 0 else '(Modified better)'}")
print(f"  - Speed Difference: {abs(orig_time - mod_time):.1f}s {'(Modified faster)' if mod_time < orig_time else '(Original faster)'}")

orig_params = sum(p.numel() for p in OriginalCNN().parameters())
mod_params = sum(p.numel() for p in ModifiedCNN().parameters())
param_reduction = ((orig_params - mod_params) / orig_params * 100)
print(f"  - Parameter Reduction: {param_reduction:.1f}%")

# ===== MY INTERPRETATION OF RESULTS =====
print(f"\n{'='*60}")
print("MY INTERPRETATION OF RESULTS")
print(f"{'='*60}")

interpretation = f"""
Based on the experimental results above, here are my observations and interpretations:

1. ACCURACY IMPACT:
   - The removal of one convolutional layer (Conv2d + ReLU + MaxPool2d) resulted in
     {accuracy_diff:.2f}% change in test accuracy.
   - {"The original model performed better, indicating that the second convolutional layer was important for feature extraction and pattern recognition." if accuracy_diff > 0 else "Surprisingly, the modified model performed better, suggesting that the original architecture might have been prone to overfitting or the second conv layer was redundant."}

2. TRAINING EFFICIENCY:
   - The modified CNN trained {'faster' if mod_time < orig_time else 'slower'} than the original, completing in {abs(orig_time - mod_time):.1f}s less time.
   - This is {'expected' if mod_time < orig_time else 'unexpected'} because {'fewer' if param_reduction > 0 else 'more'} parameters means {'less' if param_reduction > 0 else 'more'} computation per forward pass.

3. MODEL COMPLEXITY:
   - Parameter reduction of {param_reduction:.1f}% shows {'significant model simplification' if param_reduction > 0 else 'increased model complexity'}.
   - {'Fewer parameters generally mean faster training, less memory usage, and reduced risk of overfitting, but potentially lower representational capacity.' if param_reduction > 0 else 'More parameters allow for greater representational capacity but increase computational cost and overfitting risk.'}

4. FEATURE LEARNING:
   - The original CNN has two convolutional layers that can learn hierarchical features:
     * First layer: Low-level features (edges, corners)
     * Second layer: Mid-level features (shapes, textures)
   - The modified CNN relies on just one conv layer, limiting its ability to learn complex hierarchical features.
   - This explains why {'the original model achieved better accuracy - deeper feature extraction was beneficial' if accuracy_diff > 0 else 'the modified model performed similarly or better - the dataset might not require deep hierarchical features'}.

5. PRACTICAL IMPLICATIONS:
   - For applications requiring {'high accuracy' if accuracy_diff > 0 else 'computational efficiency'}, the {'original' if accuracy_diff > 0 else 'modified'} model would be preferred.
   - The trade-off between model complexity and performance is evident here.
   - In resource-constrained environments (mobile devices, embedded systems), the {'modified' if param_reduction > 0 else 'original'} model might be more suitable despite {'any accuracy loss' if accuracy_diff > 0 and param_reduction > 0 else 'increased complexity'}.

6. ARCHITECTURE INSIGHTS:
   - This experiment demonstrates that not all layers contribute equally to performance.
   - For simple datasets like CIFAR-10, {'excessive depth might lead to overfitting' if accuracy_diff < 0 else 'multiple convolutional layers help capture important patterns'}.
   - The optimal architecture depends on the complexity of the dataset and the available computational resources.

CONCLUSION:
The experiment demonstrates the importance of each layer in CNN architecture and shows
how architectural choices affect both performance and computational efficiency.
{'Removing the second convolutional layer resulted in reduced accuracy but faster training, highlighting the trade-off between model capacity and efficiency.' if accuracy_diff > 0 else 'Removing the second convolutional layer improved or maintained performance while reducing training time, suggesting the original architecture may have been unnecessarily complex for this task.'}

This analysis reinforces the importance of careful architecture design and the need to
balance model complexity with performance requirements for specific applications.
"""

print(interpretation)

print(f"\n{'='*60}")
print("EXPERIMENT COMPLETED SUCCESSFULLY")
print(f"Completed at: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
print("Based on GeeksforGeeks CNN Implementation")
print(f"{'='*60}")
```
