```python
# Assignment 5: Revised Assignment 3 with Batch Training (XOR Structure)
# kirtirajsinh Atodariya M01620157

import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Set random seed for reproducibility
torch.manual_seed(42)

print("Assignment 5: MLP Regression with Batch Training")
print("="*50)

# Load the Assignment 3 dataset (same as before)
try:
    data = pd.read_csv('/content/a3_data.csv')
    print(f"✅ Dataset loaded: {data.shape}")
except FileNotFoundError:
    print("❌ Error: a3_data.csv not found!")
    exit()

# Extract features (v1, v2, v3, v4) and target (y) - SAME AS ASSIGNMENT 3
X = torch.tensor(data[['v1', 'v2', 'v3', 'v4']].values, dtype=torch.float32)
y = torch.tensor(data['y'].values, dtype=torch.float32).unsqueeze(1)

print(f"Features shape: {X.shape}")
print(f"Target shape: {y.shape}")

# Split dataset: 75% training, 25% testing - SAME AS ASSIGNMENT 3
split_idx = int(0.75 * len(X))
X_train, X_test = X[:split_idx], X[split_idx:]
y_train, y_test = y[:split_idx], y[split_idx:]

print(f"Training samples: {len(X_train)} (75%)")
print(f"Testing samples: {len(X_test)} (25%)")

# Model - Two Hidden Layers (SAME ARCHITECTURE AS ASSIGNMENT 3)
# Following XOR code structure but for regression
model = nn.Sequential(
    nn.Linear(4, 128),    # First hidden layer: 4 → 128
    nn.ReLU(),
    nn.Linear(128, 64),   # Second hidden layer: 128 → 64
    nn.ReLU(),
    nn.Linear(64, 1)      # Output layer: 64 → 1 (regression)
)

print(f"\n✅ Model Architecture (Two Hidden Layers):")
print(f"Input layer: 4 neurons (v1, v2, v3, v4)")
print(f"Hidden layer 1: 128 neurons (ReLU)")
print(f"Hidden layer 2: 64 neurons (ReLU)")
print(f"Output layer: 1 neuron (Linear)")

# Batch setup - FOLLOWING XOR CODE STRUCTURE
batch_size = 32  # You can choose batch size
train_dataset = TensorDataset(X_train, y_train)
test_dataset = TensorDataset(X_test, y_test)
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)

print(f"✅ Batch size: {batch_size}")
print(f"Training batches: {len(train_loader)}")
print(f"Testing batches: {len(test_loader)}")

# Training setup - FOLLOWING XOR CODE STRUCTURE
loss_fn = nn.MSELoss()  # MSE for regression (same as Assignment 3)
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

print(f"\n🔥 Training with batch processing...")
print(f"Optimizer: Adam, Learning Rate: 0.001")
print(f"Loss Function: MSE (regression)")

# Training loop - BATCH TRAINING (KEY REQUIREMENT)
num_epochs = 100
train_losses = []
test_losses = []

print(f"\n{'Epoch':<6} | {'Train Loss':<12} | {'Test Loss':<12}")
print("="*35)

for epoch in range(num_epochs):
    # Training phase - BATCH PROCESSING
    model.train()
    epoch_train_loss = 0.0
    train_samples = 0

    for batch_x, batch_y in train_loader:  # BATCH TRAINING LIKE XOR
        # Forward pass
        pred = model(batch_x)
        loss = loss_fn(pred, batch_y)

        # Backward pass
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        epoch_train_loss += loss.item() * batch_x.size(0)
        train_samples += batch_x.size(0)

    avg_train_loss = epoch_train_loss / train_samples
    train_losses.append(avg_train_loss)

    # Testing phase - BATCH PROCESSING
    model.eval()
    epoch_test_loss = 0.0
    test_samples = 0

    with torch.no_grad():
        for batch_x, batch_y in test_loader:  # BATCH TESTING
            pred = model(batch_x)
            loss = loss_fn(pred, batch_y)
            epoch_test_loss += loss.item() * batch_x.size(0)
            test_samples += batch_x.size(0)

    avg_test_loss = epoch_test_loss / test_samples
    test_losses.append(avg_test_loss)

    # Print every 10 epochs (like XOR example)
    if epoch % 10 == 0 or epoch == num_epochs - 1:
        print(f"{epoch+1:<6} | {avg_train_loss:<12.6f} | {avg_test_loss:<12.6f}")

print(f"\n✅ TRAINING COMPLETED!")

# Final evaluation - FOLLOWING XOR TESTING STRUCTURE
model.eval()
with torch.no_grad():
    # Test on full datasets
    train_pred = model(X_train)
    test_pred = model(X_test)

    final_train_loss = loss_fn(train_pred, y_train)
    final_test_loss = loss_fn(test_pred, y_test)

    train_rmse = torch.sqrt(final_train_loss)
    test_rmse = torch.sqrt(final_test_loss)

print(f"\n📊 FINAL RESULTS:")
print(f"Training RMSE: {train_rmse:.6f}")
print(f"Test RMSE: {test_rmse:.6f}")

# Sample results - FOLLOWING XOR RESULTS STRUCTURE
print(f"\n📊 Sample Predictions (First 10 test samples):")
print(f"{'Sample':<8} | {'Actual':<10} | {'Predicted':<12} | {'Difference':<12}")
print("="*50)

model.eval()
with torch.no_grad():
    for i in range(min(10, len(X_test))):
        actual = y_test[i].item()
        predicted = model(X_test[i:i+1]).item()
        diff = abs(actual - predicted)
        print(f"{i+1:<8} | {actual:<10.4f} | {predicted:<12.4f} | {diff:<12.4f}")

# Plot results
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Loss plot
epochs = range(1, len(train_losses) + 1)
ax1.plot(epochs, train_losses, 'b-', label='Training Loss')
ax1.plot(epochs, test_losses, 'r--', label='Test Loss')
ax1.set_xlabel('Epoch')
ax1.set_ylabel('MSE Loss')
ax1.set_title('Training Progress with Batch Processing')
ax1.legend()
ax1.grid(True)

# Actual vs Predicted
model.eval()
with torch.no_grad():
    test_predictions = model(X_test).numpy().flatten()
    test_actual = y_test.numpy().flatten()

ax2.scatter(test_actual, test_predictions, alpha=0.6)
ax2.plot([test_actual.min(), test_actual.max()],
         [test_actual.min(), test_actual.max()], 'r--', lw=2)
ax2.set_xlabel('Actual Values')
ax2.set_ylabel('Predicted Values')
ax2.set_title('Test Set: Actual vs Predicted')
ax2.grid(True)

plt.tight_layout()
plt.show()

print(f"\n🎉 ASSIGNMENT 5 COMPLETED!")
print(f"✅ Revised Assignment 3: MLP regression model")
print(f"✅ Batch training: Using DataLoader with batch_size={batch_size}")
print(f"✅ XOR code structure: Following batch training pattern")
print(f"✅ Same dataset: a3_data.csv (v1,v2,v3,v4 → y)")
print(f"✅ Same architecture: Two hidden layers (4→128→64→1)")
print(f"✅ Final Test RMSE: {test_rmse:.6f}")
```
